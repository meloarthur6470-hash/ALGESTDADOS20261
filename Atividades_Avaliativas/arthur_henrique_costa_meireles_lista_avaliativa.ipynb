```
#18 Faça um programa que leia duas matrizes 3×3 e calcule a matriz soma e a matriz diferença, armazenando os resultados em novas listas.
matriz1 = []
matriz2 = []
print("Digite os valores da primeira matriz: ")
for i in range(3):
  linha = []
  for j in range(3):
    valor = int(input(f"Digite o valor para a posição [{i}][{j}]: "))
    linha.append(valor)

  matriz1.append(linha)

print("Digite os valores da segunda matriz: ")
for i in range(3):
  linha = []
  for j in range(3):
    valor = int(input(f"Digite o valor para a posição [{i}][{j}]: "))
    linha.append(valor)

  matriz2.append(linha)

matriz_soma=[]
matriz_diferenca = []


for i in range(3):
    linha_soma = []
    linha_diferenca = []

    for j in range(3):
        soma = matriz1[i][j] + matriz2[i][j]
        diferenca = matriz1[i][j] - matriz2[i][j]

        linha_soma.append(soma)
        linha_diferenca.append(diferenca)

    matriz_soma.append(linha_soma)
    matriz_diferenca.append(linha_diferenca)


print("Primeira matriz:")
for i in range(3):
    print(matriz1[i])

print("Segunda matriz:")
for i in range(3):
    print(matriz2[i])

print("Matriz soma:")
for i in range(3):
    print(matriz_soma[i])

print("Matriz diferença:")
for i in range(3):
    print(matriz_diferenca[i])

```

```
#19 Solicite dois conjuntos de 10 elementos fornecidos pelo usuário. Mostre a interseção, união e diferença (A − B e B − A).
conjunto_a = []
conjunto_b = []

print("Digite os elementos do primeiro conjunto:")
for i in range(10):
    elemento = int(input(f"Digite o elemento {i+1}: "))
    conjunto_a.append(elemento)

print("Digite os elementos do segundo conjunto:")
for i in range(10):
    elemento = int(input(f"Digite o elemento {i+1}: "))
    conjunto_b.append(elemento)

intersecao = []
uniao = []
diferenca_a_b = []
diferenca_b_a = []

for elemento_a in conjunto_a:
    if elemento_a in conjunto_b:
        intersecao.append(elemento_a)


for elemento_a in conjunto_a:
    if elemento_a not in uniao:
        uniao.append(elemento_a)



for elemento_b in conjunto_b:
    if elemento_b not in uniao:
        uniao.append(elemento_b)

for elemento_a in conjunto_a:
    if elemento_a not in conjunto_b:
      if elemento_a not in diferenca_a_b:
        diferenca_a_b.append(elemento_a)

for elemento_b in conjunto_b:
    if elemento_b not in conjunto_a:
      if elemento_b not in diferenca_b_a:
        diferenca_b_a.append(elemento_b)

print("Conjunto A:", conjunto_a)
print("Conjunto B:", conjunto_b)
print("Interseção:", intersecao)
print("União:", uniao)
print("Diferença (A - B):", diferenca_a_b)
print("Diferença (B - A):", diferenca_b_a)
```

```
#24 Dado um conjunto de números inteiros, crie um novo conjunto com apenas os quadrados perfeitos encontrados.
numeros = []
quadrados_perfeitos = []

quantidade = int(input("Quantos números você quer digitar? "))

for i in range(quantidade):
    numero = int(input("Digite um número: "))

    if numero not in numeros:
        numeros.append(numero)

for numero in numeros:
    raiz = 0

    while raiz * raiz <= numero:
        if raiz * raiz == numero:
            quadrados_perfeitos.append(numero)
        raiz = raiz + 1

print("Conjunto original:")
print(numeros)

print("Quadrados perfeitos encontrados:")
print(quadrados_perfeitos)
```

```
#25 Crie um dicionário que armazene o nome e o salário de 5 funcionários. Calcule o reajuste de 10% para salários inferiores a R$ 2.000 e atualize o dicionário. (Use estruturas condicionais.)
funcionarios = {}

for i in range(5):
    nome = input("Digite o nome do funcionário: ")
    salario = float(input("Digite o salário: "))

    funcionarios[nome] = salario

for nome in funcionarios:
    if funcionarios[nome] < 2000:
        funcionarios[nome] = funcionarios[nome] + (funcionarios[nome] * 0.10)

print("Funcionários com salários atualizados:")

for nome in funcionarios:
    print(nome, "- R$", funcionarios[nome])
```

```
#29 Crie um dicionário onde as chaves são nomes de produtos e os valores são listas com preços praticados em três lojas diferentes. Encontre o produto com maior variação de preço.
produtos = {}

for i in range(5):
    nome = input("Digite o nome do produto: ")

    precos = []

    for j in range(3):
        preco = float(input("Digite o preço da loja: "))
        precos.append(preco)

    produtos[nome] = precos


maior_variacao = 0
produto_maior_variacao = ""

for produto in produtos:
    maior_preco = produtos[produto][0]
    menor_preco = produtos[produto][0]

    for preco in produtos[produto]:
        if preco > maior_preco:
            maior_preco = preco

        if preco < menor_preco:
            menor_preco = preco

    variacao = maior_preco - menor_preco

    if variacao > maior_variacao:
        maior_variacao = variacao
        produto_maior_variacao = produto


print("Produtos e preços:")

for produto in produtos:
    print(produto, ":", produtos[produto])

print("Produto com maior variação de preço:")
print(produto_maior_variacao)

print("Maior variação:")
print(maior_variacao)
```

```
#30 Implemente um programa que leia uma frase e armazene em um dicionário o número de ocorrências de cada palavra (ignorando maiúsculas/minúsculas). (Use estruturas de repetição.)
frase = input("Digite uma frase: ")

frase = frase.lower()

palavras = frase.split()

ocorrencias = {}

for palavra in palavras:
    if palavra in ocorrencias:
        ocorrencias[palavra] = ocorrencias[palavra] + 1
    else:
        ocorrencias[palavra] = 1

print("ocorrências de cada palavra:")

for palavra in ocorrencias:
    print(palavra, ":", ocorrencias[palavra])
```
