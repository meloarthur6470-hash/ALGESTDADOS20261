```
#1 Crie uma lista com 15 números inteiros digitados pelo usuário. Em seguida, remova todos os números repetidos, mantendo a ordem original.
lista=[]
sem_repetidos=[]
for i in range(15):
    numero= input(f'digite o numero: ')

    lista.append(numero)
for numero in lista:
    if numero not in sem_repetidos:
        sem_repetidos.append(numero)
print(sem_repetidos)
print(lista)
```
