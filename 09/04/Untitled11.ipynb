```python
#2 Em sistemas de validação de dados numéricos, identificar se um valor é par ou ímpar é uma operação fundamental.
#Desenvolva um programa em Python que leia um número inteiro fornecido pelo usuário e informe sua paridade.]
#Crie a função eh_par(numero) que retorne True se o número for divisível por 2 e False caso contrário.
#O programa deve exibir o resultado de forma clara.
def eh_par(numero):
    if numero % 2 == 0:
        return True
    else:
        return False


numero = int(input("Digite um número inteiro: "))

if eh_par(numero):
    print(f"O número {numero} é par.")
else:
    print(f"O número {numero} é ímpar.")




```

```python
#13 Em análise de dados, determinar os valores extremos de um conjunto é operação essencial. Crie a função maior_menor(vetor) que receba uma lista de 8 números inteiros e retorne uma tupla no formato (maior, menor), sem utilizar as funções built-in max() e min().
def maior_menor(vetor):
    maior = vetor[0]
    menor = vetor[0]
    for numero in vetor:
        if numero > maior:
            maior = numero
        if numero < menor:
            menor = numero
    return (maior, menor)
```
