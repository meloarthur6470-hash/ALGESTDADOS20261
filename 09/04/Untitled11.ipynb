```python
#13 Em análise de dados, determinar os valores extremos de um conjunto é operação essencial. Crie a função maior_menor(vetor) que receba uma lista de 8 números inteiros e retorne uma tupla no formato (maior, menor), sem utilizar as funções built-in max() e min().
def maior_menor(vetor):
    maior = vetor[0]
    menor = vetor[0]
    for numero in vetor:
        if numero > maior:
            maior = numero
        if numero < menor:
            menor = numero
    return (maior, menor)
```

```python
#15 Em estatística aplicada a sistemas computacionais, o cálculo da média é operação central. Crie a função media_valores(lista) que receba uma lista de números reais e retorne a média aritmética com duas casas decimais. A função deve tratar o caso de lista vazia, retornando 0. Teste com 6 elementos fornecidos pelo usuário.
def media_valores(lista):
    if len(lista) == 0:
        return 0

    soma = sum(lista)
    media = soma / len(lista)

    return round(media, 2)


valores = []

for i in range(6):
    numero = float(input(f"Digite o {i+1}º valor: "))
    valores.append(numero)

print("Média:", media_valores(valores))
```
