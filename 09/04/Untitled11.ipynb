```python
#1 Em um sistema de controle de acesso, é necessário verificar se um valor ultrapassa um limite definido. Elabore um programa em Python que solicite ao usuário a entrada de um número inteiro e determine se ele é maior que 10. Implemente a função verificar_maior_que_dez(numero), que receba um inteiro e retorne um valor booleano. O programa deve exibir uma mensagem descritiva ao usuário com base no resultado.
def maior_que_10(numero):
    if numero > 10:
        return True
    else:
        return False
valor = int(input("digite um numero :"))
if maior_que_10(valor):
    print("o numero é maior que 10")
else:
    print("o numero não é maior que 10")
```

```python
#2 Em sistemas de validação de dados numéricos, identificar se um valor é par ou ímpar é uma operação fundamental.
#Desenvolva um programa em Python que leia um número inteiro fornecido pelo usuário e informe sua paridade.]
#Crie a função eh_par(numero) que retorne True se o número for divisível por 2 e False caso contrário.
#O programa deve exibir o resultado de forma clara.
def eh_par(numero):
    if numero % 2 == 0:
        return True
    else:
        return False


numero = int(input("Digite um número inteiro: "))

if eh_par(numero):
    print(f"O número {numero} é par.")
else:
    print(f"O número {numero} é ímpar.")




```

```python
#3 Em um sistema de cadastro, regras de validação de nomes são aplicadas automaticamente.
#Escreva um programa que leia um nome informado pelo usuário e verifique se ele contém a letra "a" (maiúscula ou minúscula).
#Implemente a função contem_letra_a(nome) que retorne um valor booleano conforme a condição.
#Exiba uma mensagem adequada ao usuário.

def contem_letra_a(nome):
    return "a" in nome.lower()

nome = input("Digite seu nome: ")

if contem_letra_a(nome):
    print("Seu nome contém a letra 'a'.")
else:
    print("Seu nome não contém a letra 'a'.")
```
